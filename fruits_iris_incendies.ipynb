Ces fonctions sont toutes déjà définies dans le module ``numpy``. Le résultat pourra se présenter sous la forme d'un tableau (pas aussi beau, que là-dessous, là c'est du markdown):

| Attribut | min | max | moyenne | std |
| --- | --- | --- | --- | --- |
| sepal_length | 4.30 | 7.90 | 5.84 | 0.83 |
| sepal_width | 2.00 |4.40 | 3.06 | 0.43 |
| petal_length | 1.00 | 6.90 | 3.76 | 1.76 |

```python
def display_stat(dico):
    dico_stat = { "min": [],"max": [], "moyenne": [], "std": []}
    for key in dico.keys():
        dico_stat["min"].append(dico[key].min())
        dico_stat["max"].append(dico[key].max())
        dico_stat["moyenne"].append(dico[key].mean())
        dico_stat["std"].append(dico[key].std())

    print(dico_stat)

display_stat(dico)
```

On donne la première normalisation (MinMax), pour aider la prise en main des fonctionnalités pandas:

```python
#code modifié par IA simplement pour supprimer les alertes de type 'futureWarning'
def normaMinMax(X):
    Xn = X.copy() # on duplique pour éviter de modifier le paramètre X
    numeric_cols = Xn.drop(columns="Fruit").columns

    # Convertir les colonnes numériques à normaliser en float pour éviter les FutureWarnings
    Xn[numeric_cols] = Xn[numeric_cols].astype(float)

    # Normaliser chaque colonne numériquement en utilisant des opérations vectorisées
    for col in numeric_cols:
        min_val = Xn[col].min()
        max_val = Xn[col].max()
        if (max_val - min_val) != 0:
            Xn[col] = (Xn[col] - min_val) / (max_val - min_val)
        else:
            Xn[col] = 0.0  # Gérer le cas où la plage est zéro

    Xn["Fruit"] = X["Fruit"]
    return Xn
```

Indépendamment de ``pandas``, la bibliothèque ``sklearn`` intègre le module ``preprocessing``, qui permet de réaliser un grand nombre de normalisations. Nous les utiliserons dans ce cours, mais pas aujourd'hui.

```python
#code modifié par IA simplement pour supprimer les alertes de type 'futureWarning'
def normaMean(X):
    X_features = X.drop(columns="Fruit")
    Xn = X_features.copy()

    # Convertir les colonnes numériques à normaliser en float pour éviter les FutureWarnings
    numeric_cols = Xn.columns
    Xn[numeric_cols] = Xn[numeric_cols].astype(float)

    for nom_colonne in Xn.columns: # On normalise les colonnes
        col_mean = X_features[nom_colonne].mean()
        col_min = X_features[nom_colonne].min()
        col_max = X_features[nom_colonne].max()
        if col_max == col_min:
            Xn[nom_colonne] = 0.0  #cas ou il n'y a pas de variance
        else:
            Xn[nom_colonne] = (X_features[nom_colonne] - col_mean) / (col_max - col_min) # Correction: division par l'étendue

    Xn["Fruit"] = X["Fruit"]
    return Xn

def normaStandard(X):
    X_features = X.drop(columns="Fruit")
    Xn = X_features.copy()

    # Convertir les colonnes numériques à normaliser en float pour éviter les FutureWarnings
    numeric_cols = Xn.columns
    Xn[numeric_cols] = Xn[numeric_cols].astype(float)

    for nom_colonne in Xn.columns: # On normalise les colonnes
        col_mean = X_features[nom_colonne].mean()
        col_std = X_features[nom_colonne].std()
        if col_std == 0:
            Xn[nom_colonne] = 0.0
        else:
            Xn[nom_colonne] = (X_features[nom_colonne] - col_mean) / col_std

    Xn["Fruit"] = X["Fruit"]
    return Xn

from numpy import linalg as LA

def normaProjBall(X):
    # Assurer que X_features a aussi les types float pour les calculs
    X_features = X.drop(columns="Fruit").astype(float) # <-- Convertir X_features en float ici
    Xn = X_features.copy() # Xn sera aussi float dès le départ

    # on normalise
    for i in range(len(Xn)):
        actuel = X_features.iloc[i, :]
        norme = (actuel**2).sum()**0.5

        if norme != 0:
            Xn.iloc[i, :] = actuel / norme
        else:
            Xn.iloc[i, :] = 0.0  # Si norm vaut zéro

    Xn["Fruit"] = X["Fruit"]
    return Xn
```

```python
import geopandas as gpd

url = "https://raw.githubusercontent.com/gregoiredavid/france-geojson/master/departements.geojson"

# A completer
departements = gpd.read_file(url)# CHARGER LES DONNEES disponible à cet url, via une fonction de lecture de geopandas

# A completer. On pourra utiliser la fonction groupby de pandas puis selectionner la bonne colonne sur laquelle calculer la moyenne
surfaces = df_incendies.groupby("Département")["Surface parcourue (m2)"].mean() # LISTE DES MOYENNES DE SURFACES PARCOURUES, PAR DEPARTEMENT

# A completer pour que les codes des départements soient bien des chaînes (traiter la Corse...)
departements["code"] = departements["code"].astype(str) #

# A completer en utilisant la fonction merge de geopandas, pour associer les départements et leur surface de feux
carte = departements.merge(surfaces, left_on="code", right_index=True, how="left") # ASSOCIER DEPARTEMENTS ET SURFACES DE FEUX, pour préparer l'affichage
centres = carte.geometry.centroid # CALCULER LE CENTROIDE DE CHAQUE DEPARTEMENT DE LA CARTE (cf package geometry de geopandas)

#
# Tracer la carte
#

fig, ax = plt.subplots(figsize=(12, 12))

# A completer : afficher les contours de la carte de France et des départements
carte.plot(ax=ax, color="whitesmoke", edgecolor="grey")

# A completer, adapter
surfaces_affichees = carte["Surface parcourue (m2)"].fillna(0)
ax.scatter(
    centres.x,
    centres.y,
    s=surfaces_affichees / surfaces_affichees.max() * 800,  # les surfaces de la carte à afficher, à traduire en surface du point affiché
    color="red",
    alpha=0.5,
    edgecolor="darkred",
    linewidth=0.5,
)

# A completer, adapter
ax.set_title("Surface moyenne brûlée par département (2023-2025)", fontsize=14)
ax.set_axis_off()
plt.show()
```

8. Le code ci-dessous permet de tracer un nuage de points des natures de feux pour chaque département. Etudiez le code, puis produisez le même style de graphique, mais sans les valeurs manquantes.

```python
import matplotlib.pyplot as plt

# Remplacer les valeurs manquantes par "Inconnue"
dfi = df_incendies.copy()
dfi["Nature"] = dfi["Nature"].fillna("Inconnue")
dfi["Département"] = dfi["Département"].astype(str)

# Nombre de feux par département et origine
comptage = ( dfi.groupby(["Département", "Nature"]).size().reset_index(name="nombre_feux") )

# Nombre total de feux par département
total = ( dfi.groupby("Département").size().reset_index(name="total_feux") )

# Ajouter le total aux donnees par dpt et origine
comptage = comptage.merge(total, on="Département")

# Calcul du pourcentage
comptage["pourcentage"] = ( comptage["nombre_feux"] / comptage["total_feux"] * 100 )

# Rendu graphique
plt.figure(figsize=(16, 7))

for origine in comptage["Nature"].unique():
    donnees = comptage[comptage["Nature"] == origine]

    plt.scatter(
        donnees["Département"],
        donnees["pourcentage"],
        label=origine,
        s=40,
        alpha=0.7
    )

plt.xlabel("Département")
plt.ylabel("Pourcentage des départs de feux (%)")
plt.title("Origine des feux par département")
plt.legend(title="Nature des feux")
plt.xticks(rotation=90)
plt.ylim(0, 100)

plt.show()
```

```python
# A vous

dfi_sans_manquantes = df_incendies.dropna(subset=["Nature"]).copy()
dfi_sans_manquantes["Département"] = dfi_sans_manquantes["Département"].astype(str)

comptage2 = (
    dfi_sans_manquantes.groupby(["Département", "Nature"]).size().reset_index(name="nombre_feux")
)
total2 = dfi_sans_manquantes.groupby("Département").size().reset_index(name="total_feux")
comptage2 = comptage2.merge(total2, on="Département")
comptage2["pourcentage"] = comptage2["nombre_feux"] / comptage2["total_feux"] * 100

plt.figure(figsize=(16, 7))
for origine in comptage2["Nature"].unique():
    donnees = comptage2[comptage2["Nature"] == origine]
    plt.scatter(donnees["Département"], donnees["pourcentage"], label=origine, s=40, alpha=0.7)

plt.xlabel("Département")
plt.ylabel("Pourcentage des départs de feux (%)")
plt.title("Origine des feux par département (valeurs manquantes exclues)")
plt.legend(title="Nature des feux")
plt.xticks(rotation=90)
plt.ylim(0, 100)
plt.show()
```

9. On cherche un graphique de base plus facile pour comparer graphiquement les natures des feux par département. Pour cela, on utilise la fonction ``pivot`` qui permet de calculer des pourcentages (ou autre) entre deux variables, puis un diagramme en barres pour représenter ces pourcentages. Ecrire le code pour le produire (avec ou sans les valeurs manquantes).


```python
# A vous

dfi_pivot = df_incendies.copy()
dfi_pivot["Nature"] = dfi_pivot["Nature"].fillna("Inconnue")

# table pivot : pour chaque département, le nombre de feux par nature
pivot = pd.pivot_table(
    dfi_pivot, index="Département", columns="Nature",
    values="Numéro", aggfunc="count", fill_value=0,
)

# on transforme en pourcentages (chaque ligne represente un departement)
pivot_pct = pivot.div(pivot.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(18, 7))
pivot_pct.plot(kind="bar", stacked=True, ax=ax, width=0.9)
ax.set_ylabel("Pourcentage des départs de feux (%)")
ax.set_title("Répartition (%) des natures de feux par département")
ax.legend(title="Nature des feux", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.show()
```
