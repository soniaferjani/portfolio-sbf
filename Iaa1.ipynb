**Question.** Ecrire une fonction ``display_stat`` qui prend en entrée un dictionnaire avec les clés de la question précédente et renvoie les statistiques 'moyenne', 'valeur max', 'valeur min', et 'déviation standard' pour chacun des 4 attributs. Ces fonctions sont toutes déjà définies dans le module ``numpy``. Le résultat pourra se présenter sous la forme d'un tableau (pas aussi beau, que là-dessous, là c'est du markdown):

| Attribut | min | max | moyenne | std |
| --- | --- | --- | --- | --- |
| sepal_length | 4.30 | 7.90 | 5.84 | 0.83 |
| sepal_width | 2.00 |4.40 | 3.06 | 0.43 |
| petal_length | 1.00 | 6.90 | 3.76 | 1.76 |
| petal_width | 0.10 | 2.50 | 1.20 | 0.76 |

```python
import numpy as np
# A vous (sans utiliser pandas vu en cours!)

def display_stat(dico):
    """Prend un dictionnaire {nom_attribut: liste_de_valeurs} et affiche/retourne
    moyenne, min, max et écart-type pour chaque attribut."""
    stats = {}
    for cle, valeurs in dico.items():
        valeurs = np.array(valeurs)
        stats[cle] = {
            "min": np.min(valeurs),
            "max": np.max(valeurs),
            "moyenne": np.mean(valeurs),
            "std": np.std(valeurs),
        }

    print(f"{'Attribut':20s} | {'min':>6s} | {'max':>6s} | {'moyenne':>8s} | {'std':>6s}")
    for cle, s in stats.items():
        print(f"{cle:20s} | {s['min']:6.2f} | {s['max']:6.2f} | {s['moyenne']:8.2f} | {s['std']:6.2f}")

    return stats

_ = display_stat(dico)
```

**Question.** Pour chaque type de normalisation vu en cours, écrire le code d'une fonction qui prend en paramètre la matrice de données initiale, et retourne une copie normalisée de cette matrice. **Attention**, la dernière colonne indique la catégorie (classe du fruit, pomme ou banane: cette colonne ne doit évidemment pas être prise en compte dans les normalisations).

On donne la première normalisation (MinMax), pour aider la prise en main des fonctionnalités pandas:

```python
# X est un dataframe pandas
def normaMinMax(X):
    Xn = X.copy() # on duplique pour éviter de modifier le paramètre X
    Xn.drop(columns="Fruit") # pour ne pas tenir compte de cette information
    nbl, nbc = Xn.shape      # les dimensions du tableau de donnees
    for i in range(nbl):     # on normalise chaque exemple, ligne
        for j in range(nbc): # on normalise chaque variable de l'exemple, colonne
            # comprendre ces lignes absolument, sinon demander au prof
            Xn.iloc[i,j] = ( X.iloc[i,j] - X.iloc[:,j].min() ) / (X.iloc[:,j].max() - X.iloc[:,j].min())
    Xn["Fruit"] = X["Fruit"]
    return Xn
```

Indépendamment de ``pandas``, la bibliothèque ``sklearn`` intègre le module ``preprocessing``, qui permet de réaliser un grand nombre de normalisations. Nous les utiliserons dans ce cours, mais pas aujourd'hui.

```python
# A vous

def normaMean(X):
    # Normalisation "centrée-réduite par l'étendue" : on centre sur la moyenne
    # et on divise par l'étendue (max - min), colonne par colonne (sauf Fruit)
    Xn = X.copy()
    colonnes = [c for c in X.columns if c != "Fruit"]
    for c in colonnes:
        Xn[c] = (X[c] - X[c].mean()) / (X[c].max() - X[c].min())
    return Xn

def normaStandard(X):
    # Normalisation standard (z-score) : on centre sur la moyenne et on
    # divise par l'écart-type, colonne par colonne (sauf Fruit)
    Xn = X.copy()
    colonnes = [c for c in X.columns if c != "Fruit"]
    for c in colonnes:
        Xn[c] = (X[c] - X[c].mean()) / X[c].std()
    return Xn

from numpy import linalg as LA

def normaProjBall(X):
    # Projection de chaque exemple sur la boule unité (norme euclidienne = 1)
    Xn = X.copy()
    colonnes = [c for c in X.columns if c != "Fruit"]
    M = X[colonnes].values.astype(float)
    normes = LA.norm(M, axis=1, keepdims=True)
    normes[normes == 0] = 1        # évite une division par zéro
    M_normalise = M / normes
    for k, c in enumerate(colonnes):
        Xn[c] = M_normalise[:, k]
    return Xn
```

**Question.** Pour chaque type de normalisation ci-avant, produire le nuage de points ainsi normalisé.

```python
# A vous

normalisations = {
    "Données brutes": fruits,
    "MinMax": normaMinMax(fruits),
    "Centrée/étendue (Mean)": normaMean(fruits),
    "Standard (z-score)": normaStandard(fruits),
    "Projection boule unité": normaProjBall(fruits),
}

fig, axes = plt.subplots(1, len(normalisations), figsize=(22, 4))

for ax, (titre, Xn) in zip(axes, normalisations.items()):
    ax.scatter(Xn["Weight"], Xn["Length"], c=Xn["Fruit"], cmap="coolwarm", s=15)
    ax.set_title(titre)
    ax.set_xlabel("Weight")
    ax.set_ylabel("Length")

plt.tight_layout()
plt.show()

# Constat :
# - MinMax et Mean conservent la forme globale du nuage (transformation affine
#   par colonne) mais changent l'échelle des axes (tout est ramené entre 0 et 1,
#   ou centré autour de 0) : les distances relatives entre points ne sont pas
#   conservées si les deux variables n'ont pas la même dispersion au départ.
# - Standard (z-score) recentre et remet chaque variable à la même échelle
#   (écart-type=1), ce qui est utile quand les variables ont des unités/échelles
#   très différentes (ici Weight varie sur ~100 alors que Length varie sur ~15).
# - ProjBall change complètement la topologie du nuage : tous les points sont
#   projetés sur le cercle unité, on ne conserve que l'information angulaire
#   (le rapport Weight/Length), la notion de distance d'origine est perdue.
```

6. En utilisant la bibliothèque geopandas (via ``!pip install geopandas`` dans une cellule), nous pouvons manipuler l'affichage de données sur une carte.

La liste des départements français est disponible au téléchargement à l'URL "https://raw.githubusercontent.com/gregoiredavid/france-geojson/master/departements.geojson". Au delà du site officiel, une documentation et des tutos intéressants sont ici: https://juliedjidji.github.io/memocarto/gallery/index.html.

Etudier la documentation de ``geopandas``, et l'utiliser pour produire une carte de France avec, pour chaque département, un cercle rouge dont la surface est proportionnelle à la somme de surface brûlées entre les années 2023 et 2025. Nous donnons ci-après le schéma général du programme, et il s'agit de produire/compléter le code pour qu'il utilise les bonnes données.

```python
import geopandas as gpd

url = "https://raw.githubusercontent.com/gregoiredavid/france-geojson/master/departements.geojson"

# A completer
departements = gpd.read_file(url)   # CHARGER LES DONNEES disponible à cet url, via une fonction de lecture de geopandas

# A completer. On pourra utiliser la fonction groupby de pandas puis selectionner la bonne colonne sur laquelle calculer la moyenne
surfaces = incendies2325.groupby("Département")["Surface parcourue (m2)"].mean()   # LISTE DES MOYENNES DE SURFACES PARCOURUES, PAR DEPARTEMENT

# A completer pour que les codes des départements soient bien des chaînes (traiter la Corse...)
departements["code"] = departements["code"].astype(str)
# le geojson distingue "2A" (Corse-du-Sud) et "2B" (Haute-Corse), déjà sous forme de code -> rien à forcer de plus,
# mais si le geojson fournissait un code numérique unique "20" pour la Corse il faudrait le splitter ici via le nom.

# A completer en utilisant la fonction merge de geopandas, pour associer les départements et leur surface de feux
carte = departements.merge(surfaces, left_on="code", right_index=True, how="left")   # ASSOCIER DEPARTEMENTS ET SURFACES DE FEUX, pour préparer l'affichage
centres = carte.geometry.centroid   # CALCULER LE CENTROIDE DE CHAQUE DEPARTEMENT DE LA CARTE (cf package geometry de geopandas)

#
# Tracer la carte
#

fig, ax = plt.subplots(figsize=(12, 12))

# A completer : afficher les contours de la carte de France et des départements
carte.plot(ax=ax, color="whitesmoke", edgecolor="grey")

# A completer, adapter
surfaces_affichees = carte["Surface parcourue (m2)"].fillna(0)
ax.scatter(
    centres.x,
    centres.y,
    s=surfaces_affichees / surfaces_affichees.max() * 800,  # les surfaces de la carte à afficher, à traduire en surface du point affiché
    color="red",
    alpha=0.5,
    edgecolor="darkred",
    linewidth=0.5,
)

# A completer, adapter
ax.set_title("Surface moyenne brûlée par département (2023-2025)", fontsize=14)
ax.set_axis_off()
plt.show()

# NB : cette cellule nécessite un accès internet (téléchargement du geojson et
# éventuellement `!pip install geopandas` au préalable) : elle n'a pas pu être
# exécutée dans cet environnement, mais la logique/le code est prêt à tourner.
```

8. Le code ci-dessous permet de tracer un nuage de points des natures de feux pour chaque département. Etudiez le code, puis produisez le même style de graphique, mais sans les valeurs manquantes.

```python
import matplotlib.pyplot as plt

# Remplacer les valeurs manquantes par "Inconnue"
dfi = incendies2325.copy()
dfi["Nature"] = dfi["Nature"].fillna("Inconnue")
dfi["Département"] = dfi["Département"].astype(str)

# Nombre de feux par département et origine
comptage = ( dfi.groupby(["Département", "Nature"]).size().reset_index(name="nombre_feux") )

# Nombre total de feux par département
total = ( dfi.groupby("Département").size().reset_index(name="total_feux") )

# Ajouter le total aux donnees par dpt et origine
comptage = comptage.merge(total, on="Département")

# Calcul du pourcentage
comptage["pourcentage"] = ( comptage["nombre_feux"] / comptage["total_feux"] * 100 )

# Rendu graphique
plt.figure(figsize=(16, 7))

for origine in comptage["Nature"].unique():
    donnees = comptage[comptage["Nature"] == origine]

    plt.scatter(
        donnees["Département"],
        donnees["pourcentage"],
        label=origine,
        s=40,
        alpha=0.7
    )

plt.xlabel("Département")
plt.ylabel("Pourcentage des départs de feux (%)")
plt.title("Origine des feux par département")
plt.legend(title="Nature des feux")
plt.xticks(rotation=90)
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

9. On cherche un graphique de base plus facile pour comparer graphiquement les natures des feux par département. Pour cela, on utilise la fonction ``pivot`` qui permet de calculer des pourcentages (ou autre) entre deux variables, puis un diagramme en barres pour représenter ces pourcentages. Ecrire le code pour le produire (avec ou sans les valeurs manquantes).
10. Commmentez le diagramme obtenu.

```python
# A vous

dfi_pivot = incendies2325.copy()
dfi_pivot["Nature"] = dfi_pivot["Nature"].fillna("Inconnue")

# table pivot : pour chaque département, le nombre de feux par nature
pivot = pd.pivot_table(
    dfi_pivot, index="Département", columns="Nature",
    values="Numéro", aggfunc="count", fill_value=0,
)

# on transforme en pourcentages (par ligne, donc par département)
pivot_pct = pivot.div(pivot.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(18, 7))
pivot_pct.plot(kind="bar", stacked=True, ax=ax, width=0.9)
ax.set_ylabel("Pourcentage des départs de feux (%)")
ax.set_title("Répartition (%) des natures de feux par département")
ax.legend(title="Nature des feux", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()

# 10. Commentaire :
# Ce diagramme en barres empilées est beaucoup plus lisible que le nuage de
# points précédent pour comparer les départements entre eux : on voit
# immédiatement, pour chaque département, la proportion de chaque nature de
# feu (d'un coup d'œil sur la hauteur du segment de couleur), alors que le
# nuage de points obligeait à comparer visuellement des points isolés le long
# de chaque colonne. On distingue par exemple des départements où la
# malveillance domine largement, et d'autres où les causes accidentelles /
# involontaires sont majoritaires.
```
